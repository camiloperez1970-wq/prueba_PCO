```python
import sagemaker
from sagemaker.model import Model
from sagemaker import image_uris, get_execution_role
import time

# 1. Configuración inicial
role = get_execution_role()
sagemaker_session = sagemaker.Session()
region = sagemaker_session.boto_region_name

# Define el bucket y la ruta de S3
bucket = "datospco"
model_key = "modelo.tar.gz"
model_s3_uri = f"s3://{bucket}/{model_key}"

# 2. Obtén la URI de la imagen de scikit-learn 1.6-1
container_image = image_uris.retrieve(
    framework="sklearn",
    region=region,
    version="1.4-1"
)

# 3. Crea el objeto del modelo de SageMaker
sagemaker_model = Model(
    image_uri=container_image,
    model_data=model_s3_uri,
    role=role,
    entry_point="inference2.py",
    sagemaker_session=sagemaker_session
)

# 4. Despliega el modelo en un endpoint
endpoint_name = f"modelo-sklearn-1-2-{int(time.time())}"

print(f"Desplegando el modelo en el endpoint: {endpoint_name}")

predictor = sagemaker_model.deploy(
    initial_instance_count=1,
    instance_type="ml.t2.medium",
    endpoint_name=endpoint_name
)

print("\n¡Despliegue completado! El endpoint está listo para recibir peticiones.")
print(f"Endpoint en vivo: {predictor.endpoint_name}")
```

```python
from sagemaker.sklearn.model import SKLearnModel

sk_model = SKLearnModel(
    model_data=model_s3_uri,
    role=role,
    entry_point="inference2.py",
    source_dir=".",    # donde está requirements.txt
    dependencies=["requirements.txt"],
    framework_version="1.2-1",  # misma versión que usaste
    sagemaker_session=sagemaker_session
)

endpoint_name = f"modelo-sklearn-1-2-{int(time.time())}"

predictor = sk_model.deploy(
    initial_instance_count=1,
    instance_type="ml.t2.medium",
    endpoint_name=endpoint_name
)
```

```python
import sagemaker
from sagemaker.sklearn.model import SKLearnModel
import time

role = sagemaker.get_execution_role()
sagemaker_session = sagemaker.Session()

bucket = "datospco"
model_key = "modelRF.tar.gz"
model_s3_uri = f"s3://{bucket}/{model_key}"

sk_model = SKLearnModel(
    model_data=model_s3_uri,
    role=role,
    entry_point="inference2.py",
    source_dir=".",
    framework_version="1.2-1",  # versión compatible con scikit-learn
    py_version="py3",
    sagemaker_session=sagemaker_session
)

endpoint_name = f"rf-endpoint-{int(time.time())}"
predictor = sk_model.deploy(
    initial_instance_count=1,
    instance_type="ml.t2.medium",
    endpoint_name=endpoint_name
)

print(f"✅ Endpoint listo: {predictor.endpoint_name}")
```
