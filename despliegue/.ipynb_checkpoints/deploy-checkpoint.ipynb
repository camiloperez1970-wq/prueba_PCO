```python
import tarfile

# Ruta al archivo tar.gz
ruta = "../modelo.tar.gz"

with tarfile.open(ruta, "r:gz") as tar:
    # Mostrar el contenido
    print("Archivos dentro del tar.gz:")
    tar.list()

    # Supongamos que quieres leer un archivo .txt llamado "datos.txt"
    member = tar.getmember("requirements.txt")
    f = tar.extractfile(member)
    if f:
        contenido = f.read().decode("utf-8")  # Decodificar a texto
        print("Contenido del txt:\n", contenido)
```

```python
import sagemaker
from sagemaker.model import Model
from sagemaker.predictor import Predictor
from sagemaker import image_uris, get_execution_role

# Obtén el rol de ejecución de tu notebook
sagemaker_session = sagemaker.Session()
role = get_execution_role()
```

```python
# Define la ruta S3 del modelo ya entrenado
model_data_uri = 's3://datospco/modelo.tar.gz'

# Define la imagen de inferencia. La URL de la imagen varía según el framework.
# Por ejemplo, para scikit-learn:
container_image = image_uris.retrieve(
    framework='sklearn',
    region=sagemaker_session.boto_region_name,
    version='1.2-1' # Usa una versión compatible con tu modelo
)
```

```python
# Crea el objeto del modelo de SageMaker
sagemaker_model = Model(
    image_uri=container_image,
    model_data=model_data_uri,
    role=role,
    predictor_cls=Predictor,
    sagemaker_session=sagemaker_session
)
```

```python
# Despliega el modelo en un endpoint.
# Esto puede tardar varios minutos.
endpoint_name = 'endpoint-xgboost2'
predictor = sagemaker_model.deploy(
    initial_instance_count=1,
    instance_type='ml.t2.medium', # Tipo de instancia para el endpoint
    endpoint_name=endpoint_name
)

print(f"Modelo desplegado en el endpoint: {endpoint_name}")
```

```python
import sagemaker
from sagemaker.sklearn.model import SKLearnModel
from sagemaker import get_execution_role
import time

# 1. Configuración inicial
# Obtiene el rol de ejecución de tu entorno de SageMaker
role = get_execution_role()
sagemaker_session = sagemaker.Session()

# Define el bucket de S3 y la ruta donde está tu modelo empaquetado
bucket = "datospco"
model_key = "modelo.tar.gz"
model_s3_uri = f"s3://{bucket}/{model_key}"

# 2. Crea el objeto del modelo de SageMaker
# SKLearnModel simplifica el despliegue para modelos de scikit-learn
sklearn_model = SKLearnModel(
    model_data=model_s3_uri,
    role=role,
    entry_point="inference2.py",  # Tu script de inferencia
    framework_version="1.6-1",   # Versión de scikit-learn
    sagemaker_session=sagemaker_session
)

# 3. Despliega el modelo en un endpoint
# Se genera un nombre único para el endpoint para evitar errores si ya existe uno
endpoint_name = f"modelo-rf-{int(time.time())}"

print(f"Desplegando el modelo como un endpoint de SageMaker con el nombre: {endpoint_name}")

# El método deploy despliega el modelo y crea un endpoint en vivo
predictor = sklearn_model.deploy(
    initial_instance_count=1,
    instance_type="ml.t2.medium",
    endpoint_name=endpoint_name
)

print("\n¡Despliegue completado! El endpoint está listo para recibir peticiones.")
print(f"Endpoint en vivo: {predictor.endpoint_name}")

# Puedes usar el objeto predictor para hacer inferencias.
# Por ejemplo, para hacer una predicción:
# result = predictor.predict([[5.1, 3.5, 1.4, 0.2]])
# print(result)
```

```python
from sagemaker.sklearn.model import SKLearnModel
import sagemaker
import time

role = sagemaker.get_execution_role()
sagemaker_session = sagemaker.Session()
region = sagemaker_session.boto_region_name

# Ruta del modelo en S3
bucket = "datospco"
model_key = "modelo.tar.gz"
model_s3_uri = f"s3://{bucket_name}/{s3_key}"

# Crear el SKLearnModel
sk_model = SKLearnModel(
    model_data=model_s3_uri,
    role=role,
    entry_point="inference2.py",
    source_dir=".",  # Aquí está requirements.txt y inference2.py
    framework_version="1.2-1",
    py_version="py3",
    sagemaker_session=sagemaker_session
)

# Desplegar el endpoint
endpoint_name = f"modelo-sklearn-xgb-{int(time.time())}"
predictor = sk_model.deploy(
    initial_instance_count=1,
    instance_type="ml.t2.medium",
    endpoint_name=endpoint_name
)

print(f"✅ Endpoint listo: {predictor.endpoint_name}")
```
