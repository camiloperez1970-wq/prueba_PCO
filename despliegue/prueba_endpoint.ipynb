```python
import boto3
import json


aws_region = 'us-east-1'
endpoint_name = 'rf-endpoint-1755551397' # Reemplaza con el nombre exacto de tu endpoint

# Crea un cliente para el servicio de SageMaker Runtime
runtime = boto3.client('sagemaker-runtime', region_name=aws_region)

# --- 2. Preparar el payload de la solicitud ---
# Aquí está el JSON que tu endpoint espera
payload = {
    "instances": [
        {
            "tipo_transaccion": "Sale",
            "valor_transaccion": 59000.0, # string
            "id_aliado": 21,                    # int
            "genero": "F",                       # string
            "estrato_social": 6,                 # int o float (coincidir con entrenamiento)
            "estado_civil": "Casado",            # string
            "ocupacion": "Ama de Casa",          # string
            "ciudad": "BOGOTA"                   # string
        }
    ]
}
# Convierte el diccionario a un string JSON
body = json.dumps(payload)

# --- 3. Invocar el endpoint ---
try:
    response = runtime.invoke_endpoint(
        EndpointName=endpoint_name,
        ContentType='application/json',
        Body=body
    )

    # --- 4. Procesar y mostrar la respuesta ---
    result = json.loads(response['Body'].read().decode('utf-8'))
    
    print("Predicción recibida del endpoint:")
    print(json.dumps(result, indent=2))

except Exception as e:
    print(f"Ocurrió un error al invocar el endpoint: {e}")
```
